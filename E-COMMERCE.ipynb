```python
#Create important business KPIs
print("Total Sales:",Retail["Sales"].sum())
print("Total Profit:",Retail["Quantity"].sum())
print("Total Customers:",Retail["Customer ID"].nunique())
print("Total Profit:",Retail["Profit"].sum())
      

```

```python
#VISUALIZATIONS
#Sales by category
    
    category_sales = (
    Retail.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

```python
 plt.figure(figsize=(8, 5))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title("Sales by Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.show()
```

```python
#Profit by category
Profit_by_category=Retail.groupby("Category")["Sales"].sum().sort_values(ascending=True)
```

```python
Profit_by_category
```

```python
sns.barplot(x=Profit_by_category.values,y=Profit_by_category.index)
plt.title("Profit_by_category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.show()
```

```python
#Customer segment analysis
segment_analysis = (
    Retail.groupby("Segment")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique"),
          Customers=("Customer ID", "nunique")
      )
      .sort_values("Sales", ascending=False)
)

segment_analysis
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=segment_analysis.reset_index(),
    x="Segment",
    y="Sales"
)

plt.title("Sales by Customer Segment")
plt.show()
```

```python
top_customers = (
    Retail.groupby("Customer Name")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
      .sort_values("Sales", ascending=False)
      .head(10)
)

top_customers
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_customers.reset_index(),
    x="Sales",
    y="Customer Name"
)

plt.title("Top 10 Customers by Sales")
plt.show()
```

```python
 # Sales vs Profit

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=Retail,
    x="Sales",
    y="Profit",
    hue="Category",
    alpha=0.6
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
#TOP 10 BOTTOM SALES BY CITYs

City_sales=Retail.groupby("City")["Profit"].sum().sort_values().tail(10)
City_sales
```

```python
plt.figure(figsize=(14,8))
sns.barplot(x=City_sales.index,y=City_sales.values)
plt.title("CITY WSIE SALES")
plt.xlabel("sales")
plt.ylabel("city")
plt.xticks(rotation=45)
plt.show()
```

```python
Retail["Order Date"] = pd.to_datetime(Retail["Order Date"])

Retail["Year"] = Retail["Order Date"].dt.year
Retail["Month"] = Retail["Order Date"].dt.to_period("M")
```

```python
monthly_sales = (
    Retail.groupby("Month")["Sales"]
      .sum()
      .reset_index()
)

monthly_sales["Month"] = monthly_sales["Month"].astype(str)
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Sales",
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
